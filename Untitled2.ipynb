```python
import numpy as np
import pandas as pd
import random
import joblib
import os
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Conv1D, Dense, Dropout, BatchNormalization,
                                     LayerNormalization, MultiHeadAttention, Masking,
                                     Bidirectional, GRU, concatenate, GlobalAveragePooling1D,
                                     GlobalMaxPooling1D, Add)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau, ModelCheckpoint
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')

# Set seed for reproducibility
np.random.seed(42)
tf.random.set_seed(42)
random.seed(42)

CLASSES = [
    "CIVILIAN_AIRCRAFT", "FIGHTER_AIRCRAFT", "HELICOPTER", "DRONE",
    "SURFACE_MISSILE", "AERIAL_MISSILE", "SKIFF", "BOAT", "SHIP"
]

# --- 1. EXTENDED PHYSICS-INFORMED FEATURE EXTRACTOR ---
def build_features(df):
    df = df.copy()
    df = df.sort_values(['track_no', 'ntp_sec']).reset_index(drop=True)

    # Time Delta
    df['dt'] = df.groupby('track_no')['ntp_sec'].diff().fillna(2.0).replace(0, 2.0)
    dt = df['dt'].values

    # Velocity Magnitudes
    df['vel_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2 + df['vel_z']**2)
    df['vel_xy_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2)
    df['range_mag'] = np.sqrt(df['pos_x']**2 + df['pos_y']**2 + df['pos_z']**2)

    # Accelerations (First Derivative of Velocity)
    df['accel_x'] = df.groupby('track_no')['vel_x'].diff() / dt
    df['accel_y'] = df.groupby('track_no')['vel_y'].diff() / dt
    df['accel_z'] = df.groupby('track_no')['vel_z'].diff() / dt
    df[['accel_x', 'accel_y', 'accel_z']] = df[['accel_x', 'accel_y', 'accel_z']].fillna(0.0)
    df['accel_mag'] = np.sqrt(df['accel_x']**2 + df['accel_y']**2 + df['accel_z']**2)

    # Jerk (Second Derivative of Velocity)
    df['jerk_x'] = df.groupby('track_no')['accel_x'].diff() / dt
    df['jerk_y'] = df.groupby('track_no')['accel_y'].diff() / dt
    df['jerk_z'] = df.groupby('track_no')['accel_z'].diff() / dt
    df[['jerk_x', 'jerk_y', 'jerk_z']] = df[['jerk_x', 'jerk_y', 'jerk_z']].fillna(0.0)
    df['jerk_mag'] = np.sqrt(df['jerk_x']**2 + df['jerk_y']**2 + df['jerk_z']**2)

    # Inertial Responsiveness Ratio
    df['jerk_to_accel_ratio'] = df['jerk_mag'] / (df['accel_mag'] + 1e-5)

    # 3D Vector Differential Geometry: Curvature (kappa) & Torsion (tau)
    v = df[['vel_x', 'vel_y', 'vel_z']].values
    a = df[['accel_x', 'accel_y', 'accel_z']].values
    j = df[['jerk_x', 'jerk_y', 'jerk_z']].values

    v_cross_a = np.cross(v, a)
    v_cross_a_mag = np.linalg.norm(v_cross_a, axis=1)
    v_mag = df['vel_mag'].values

    # Curvature = ||v x a|| / ||v||^3
    curvature = v_cross_a_mag / (v_mag**3 + 1e-6)
    df['curvature_3d'] = np.nan_to_num(curvature)

    # Torsion = ((v x a) . j) / ||v x a||^2
    dot_v_cross_a_j = np.einsum('ij,ij->i', v_cross_a, j)
    torsion = dot_v_cross_a_j / (v_cross_a_mag**2 + 1e-6)
    df['torsion_3d'] = np.nan_to_num(torsion)

    # Specific Mechanical Energy Dynamics
    g = 9.81
    df['specific_energy'] = g * df['pos_z'] + 0.5 * (df['vel_mag']**2)
    df['energy_rate'] = g * df['vel_z'] + (df['vel_x']*df['accel_x'] + df['vel_y']*df['accel_y'] + df['vel_z']*df['accel_z'])

    # Orientational & Dynamic Features
    df['z_ratio'] = df['vel_z'] / (df['vel_mag'] + 1e-8)
    df['vertical_accel_ratio'] = df['accel_z'] / (df['accel_mag'] + 1e-8)
    df['heading'] = np.degrees(np.arctan2(df['vel_y'], df['vel_x'])) % 360

    # Rolling Statistical Windows (Capturing Jitter & Transient Micro-Maneuvers)
    for col in ['accel_mag', 'jerk_mag', 'energy_rate']:
        df[f'{col}_roll_mean'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).mean())
        df[f'{col}_roll_std'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).std()).fillna(0.0)

    # Cumulative Tracking Extremes
    df['cum_max_accel'] = df.groupby('track_no')['accel_mag'].cummax()
    df['cum_max_jerk'] = df.groupby('track_no')['jerk_mag'].cummax()
    df['cum_max_energy_rate'] = df.groupby('track_no')['energy_rate'].cummax()

    return df.fillna(0.0)


# --- 2. MULTI-CLASS SIMULATION ENGINE WITH DISGUISE MANEUVERS ---
def generate_complex_trajectory(track_class, duration=60, dt=2.0):
    pos_x_list, pos_y_list, pos_z_list = [], [], []
    vel_x_list, vel_y_list, vel_z_list = [], [], []
    track_no_list, track_class_list, ntp_sec_list = [], [], []

    track_id = random.randint(100000, 999999)
    spawn_radius = random.uniform(5000, 70000)
    spawn_angle = random.uniform(0, 2 * np.pi)
    pos_x, pos_y = spawn_radius * np.cos(spawn_angle), spawn_radius * np.sin(spawn_angle)

    # Disguise / Spoofing Logic Injection
    is_disguised = False
    disguise_breakout_step = int(duration * random.uniform(0.4, 0.7))

    # 25% chance of tactical targets attempting dynamic disguise
    if track_class in ["FIGHTER_AIRCRAFT", "DRONE", "SURFACE_MISSILE"] and random.random() < 0.25:
        is_disguised = True

    # Initial State Initialization
    if track_class == "CIVILIAN_AIRCRAFT":
        pos_z, v_mag = np.random.uniform(7000, 11000), np.random.uniform(200, 260)
    elif track_class == "FIGHTER_AIRCRAFT":
        if is_disguised: # Pretend to be Civilian airliner initially
            pos_z, v_mag = np.random.uniform(8000, 10000), np.random.uniform(210, 250)
        else:
            pos_z, v_mag = np.random.uniform(500, 15000), np.random.uniform(250, 600)
    elif track_class == "HELICOPTER":
        pos_z, v_mag = np.random.uniform(50, 2000), np.random.uniform(10, 70)
    elif track_class == "DRONE":
        if is_disguised: # Pretend to be Helicopter
            pos_z, v_mag = np.random.uniform(100, 800), np.random.uniform(15, 35)
        else:
            pos_z, v_mag = np.random.uniform(20, 3000), np.random.uniform(5, 60)
    elif track_class == "SURFACE_MISSILE":
        if is_disguised: # Low sea-skimming pretend craft initial flight
            pos_z, v_mag = np.random.uniform(15, 30), np.random.uniform(150, 250)
        else:
            pos_z, v_mag = 0.0, np.random.uniform(300, 800)
    elif track_class == "AERIAL_MISSILE":
        pos_z, v_mag = np.random.uniform(2000, 12000), np.random.uniform(600, 1200)
    elif track_class == "SKIFF":
        pos_z, v_mag = 0.0, np.random.uniform(5, 25)
    elif track_class == "BOAT":
        pos_z, v_mag = 0.0, np.random.uniform(4, 18)
    elif track_class == "SHIP":
        pos_z, v_mag = 0.0, np.random.uniform(2, 10)

    heading_angle = random.uniform(0, 2 * np.pi)
    vel = np.array([v_mag * np.cos(heading_angle), v_mag * np.sin(heading_angle), 0.0])
    start_time = 1700000000

    for t in range(duration):
        # Handle Breakout Phase for Disguised Targets
        active_class = track_class
        if is_disguised and t >= disguise_breakout_step:
            # Reveal true nature via kinematic burst maneuver
            if track_class == "FIGHTER_AIRCRAFT":
                vel *= 1.8  # Supersonic breakout acceleration
                vel[2] += random.choice([-150, 150]) # Steep dive/climb
            elif track_class == "DRONE":
                heading_angle += np.radians(random.uniform(60, 120)) # Sharp non-aerodynamic turn
                vel[0], vel[1] = v_mag * np.cos(heading_angle), v_mag * np.sin(heading_angle)
            elif track_class == "SURFACE_MISSILE":
                vel[2] += 250 # Terminal vertical pop-up climb
            is_disguised = False # Fully revealed

        # Kinematic Dynamics Engine Per Class
        if active_class == "CIVILIAN_AIRCRAFT":
            vel += np.random.normal(0, 0.05, 3) # Very high stability
        elif active_class == "FIGHTER_AIRCRAFT":
            if random.random() < 0.2: # High g maneuvers
                heading_angle += np.radians(random.uniform(-12, 12))
                vel[0], vel[1] = np.linalg.norm(vel[:2]) * np.cos(heading_angle), np.linalg.norm(vel[:2]) * np.sin(heading_angle)
                vel[2] += random.uniform(-30, 30)
        elif active_class == "HELICOPTER":
            if random.random() < 0.15:
                vel[2] = random.uniform(-10, 10) # Vertical climb/descend
        elif active_class == "DRONE":
            vel += np.random.normal(0, 1.2, 3) # High frequency jitter/wind disturbance
            if random.random() < 0.25:
                heading_angle += np.radians(random.uniform(-35, 35))
                vel[0], vel[1] = np.linalg.norm(vel[:2]) * np.cos(heading_angle), np.linalg.norm(vel[:2]) * np.sin(heading_angle)
        elif active_class in ["SURFACE_MISSILE", "AERIAL_MISSILE"]:
            vel *= 1.02 # Continuous rocket motor boost
            vel += np.random.normal(0, 0.5, 3)
        elif active_class == "SKIFF":
            vel[2] = 0.0
            if random.random() < 0.3: # Sharp sea surface zig-zags
                heading_angle += np.radians(random.uniform(-25, 25))
                vel[0], vel[1] = np.linalg.norm(vel[:2]) * np.cos(heading_angle), np.linalg.norm(vel[:2]) * np.sin(heading_angle)
        elif active_class == "BOAT":
            vel[2] = 0.0
            if random.random() < 0.1:
                heading_angle += np.radians(random.uniform(-5, 5))
                vel[0], vel[1] = np.linalg.norm(vel[:2]) * np.cos(heading_angle), np.linalg.norm(vel[:2]) * np.sin(heading_angle)
        elif active_class == "SHIP":
            vel[2] = 0.0
            vel += np.random.normal(0, 0.01, 3) # Massive inertia, negligible change

        # Surface Boundary Enforcements
        if track_class in ["SKIFF", "BOAT", "SHIP"]:
            pos_z, vel[2] = 0.0, 0.0
        else:
            pos_z = max(0.0, pos_z + vel[2] * dt)

        pos_x += vel[0] * dt
        pos_y += vel[1] * dt

        # Measurement Noise Injections
        pos_x_list.append(pos_x + np.random.normal(0, 5.0))
        pos_y_list.append(pos_y + np.random.normal(0, 5.0))
        pos_z_list.append(pos_z + (0.0 if track_class in ["SKIFF", "BOAT", "SHIP"] else np.random.normal(0, 3.0)))
        vel_x_list.append(vel[0] + np.random.normal(0, 0.5))
        vel_y_list.append(vel[1] + np.random.normal(0, 0.5))
        vel_z_list.append(vel[2] + np.random.normal(0, 0.5))

        track_no_list.append(track_id)
        track_class_list.append(track_class)
        ntp_sec_list.append(start_time + int(t * dt))

    return pd.DataFrame({
        'pos_x': pos_x_list, 'pos_y': pos_y_list, 'pos_z': pos_z_list,
        'vel_x': vel_x_list, 'vel_y': vel_y_list, 'vel_z': vel_z_list,
        'track_no': track_no_list, 'track_class': track_class_list, 'ntp_sec': ntp_sec_list
    })

def generate_dataset(num_samples_per_class=200, duration=60):
    dfs = []
    for cls in CLASSES:
        for _ in range(num_samples_per_class):
            df = generate_complex_trajectory(cls, duration=duration)
            if df is not None and len(df) > 5:
                dfs.append(df)
    return pd.concat(dfs, ignore_index=True)


# --- 3. STATE-OF-THE-ART HYBRID ARCHITECTURE ---
def build_transformer_hybrid_v2(input_shape, num_classes):
    inputs = Input(shape=input_shape)
    masked = Masking(mask_value=0.0)(inputs)

    # Multi-Scale Inception CNN (Local Pattern Extraction)
    c1 = Conv1D(32, kernel_size=3, padding='same', activation='relu')(masked)
    c2 = Conv1D(32, kernel_size=5, padding='same', activation='relu')(masked)
    c3 = Conv1D(32, kernel_size=7, padding='same', activation='relu')(masked)
    x = concatenate([c1, c2, c3])
    x = BatchNormalization()(x)

    # Sequence Modeling with Residual Bi-GRU
    gru_out = Bidirectional(GRU(64, return_sequences=True, dropout=0.2))(x)
    gru_proj = Dense(128)(gru_out)
    x_proj = Dense(128)(x)
    x = Add()([gru_proj, x_proj]) # Residual Skip Connection

    # Global Temporal Context via Multi-Head Attention
    attn_out = MultiHeadAttention(num_heads=4, key_dim=32, dropout=0.1)(x, x)
    x = LayerNormalization()(x + attn_out)

    # Dual Pooling Head (Average captures general trend, Max captures sudden transient breakout)
    avg_pool = GlobalAveragePooling1D()(x)
    max_pool = GlobalMaxPooling1D()(x)
    pooled = concatenate([avg_pool, max_pool])

    # Classification Layer Head
    dense = Dense(128, activation='relu')(pooled)
    dense = BatchNormalization()(dense)
    dense = Dropout(0.4)(dense)
    dense = Dense(64, activation='relu')(dense)
    dense = BatchNormalization()(dense)
    outputs = Dense(num_classes, activation='softmax')(dense)

    model = Model(inputs, outputs)
    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    return model


# --- 4. MAIN PIPELINE EXECUTION ---
if __name__ == "__main__":
    if not os.path.exists("trained_model"):
        os.makedirs("trained_model")

    print("[1/5] Generating multi-class complex trajectory data (including disguise scenarios)...")
    df_raw = generate_dataset(num_samples_per_class=220, duration=60)

    print("[2/5] Extracting higher-order 3D differential geometry & energy features...")
    df_enhanced = build_features(df_raw)

    feature_cols = [
        'pos_z', 'vel_mag', 'vel_xy_mag', 'range_mag', 'accel_mag', 'jerk_mag',
        'jerk_to_accel_ratio', 'curvature_3d', 'torsion_3d', 'specific_energy',
        'energy_rate', 'z_ratio', 'vertical_accel_ratio',
        'accel_mag_roll_mean', 'accel_mag_roll_std',
        'jerk_mag_roll_mean', 'jerk_mag_roll_std',
        'energy_rate_roll_mean', 'energy_rate_roll_std',
        'cum_max_accel', 'cum_max_jerk', 'cum_max_energy_rate'
    ]

    print(f"Total Derived Features per Timestep: {len(feature_cols)}")

    # Fit and Save Scaler
    scaler = StandardScaler()
    df_enhanced[feature_cols] = scaler.fit_transform(df_enhanced[feature_cols])
    joblib.dump(scaler, "trained_model/feature_scaler.pkl")

    # Dynamic Sliding Sequence Generation
    seq_length = 40
    track_sequences, track_labels = [], []

    print("[3/5] Building sequence windows for early classification...")
    for track_id, group in df_enhanced.groupby('track_no'):
        data_matrix = group.sort_values('ntp_sec')[feature_cols].values
        label = group['track_class'].iloc[0]

        for i in range(1, len(data_matrix) + 1):
            if i <= seq_length:
                window = data_matrix[:i]
                padded = np.pad(window, ((seq_length - len(window), 0), (0, 0)), 'constant')
                track_sequences.append(padded)
            else:
                window = data_matrix[i-seq_length:i]
                track_sequences.append(window)
            track_labels.append(label)

    X = np.array(track_sequences)
    y = np.array(track_labels)

    le = LabelEncoder()
    y_encoded = le.fit_transform(y)
    joblib.dump(le, "trained_model/label_encoder.pkl")

    X_train, X_test, y_train, y_test = train_test_split(
        X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded
    )

    print(f"Dataset Formatted: X_train shape = {X_train.shape}, Classes = {len(le.classes_)}")

    print("[4/5] Constructing and compiling SOTA Multi-Class Radar Classifier...")
    model = build_transformer_hybrid_v2((X.shape[1], X.shape[2]), len(le.classes_))

    callbacks = [
        EarlyStopping(monitor='val_accuracy', patience=10, restore_best_weights=True),
        ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=4, min_lr=1e-5),
        ModelCheckpoint("trained_model/radar_target_classifier.keras", monitor='val_accuracy', save_best_only=True)
    ]

    print("[5/5] Training Model...")
    model.fit(
        X_train, y_train,
        validation_data=(X_test, y_test),
        epochs=30,
        batch_size=64,
        callbacks=callbacks
    )

    print("\nTraining Complete! Model saved successfully in 'trained_model/radar_target_classifier.keras'.")
```

```python
import numpy as np
import pandas as pd
import random

CLASSES = [
    "CIVILIAN_AIRCRAFT", "FIGHTER_AIRCRAFT", "HELICOPTER", "DRONE",
    "SURFACE_MISSILE", "AERIAL_MISSILE", "SKIFF", "BOAT", "SHIP"
]

def generate_stress_trajectory(track_class, duration=60, dt=10.0):
    pos_x_list, pos_y_list, pos_z_list = [], [], []
    vel_x_list, vel_y_list, vel_z_list = [], [], []
    track_no_list, track_class_list, ntp_sec_list = [], [], []

    track_id = random.randint(100000, 999999)
    spawn_radius = random.uniform(10000, 80000)
    spawn_angle = random.uniform(0, 2 * np.pi)
    pos_x, pos_y = spawn_radius * np.cos(spawn_angle), spawn_radius * np.sin(spawn_angle)

    # ---------------------------------------------------
    # 1. DISGUISE INITIATION
    # ---------------------------------------------------
    is_disguised = False
    active_class = track_class
    disguise_breakout_step = int(duration * random.uniform(0.4, 0.7))

    # 25% chance for tactical targets to spawn disguised
    if track_class in ["FIGHTER_AIRCRAFT", "DRONE", "SURFACE_MISSILE"] and random.random() < 0.25:
        is_disguised = True
        if track_class == "FIGHTER_AIRCRAFT": active_class = "CIVILIAN_AIRCRAFT"
        elif track_class == "DRONE": active_class = "HELICOPTER"
        elif track_class == "SURFACE_MISSILE": active_class = "BOAT"

    # ---------------------------------------------------
    # 2. BOUNDARY INITIALIZATIONS (Based on Active Class)
    # ---------------------------------------------------
    if active_class == "CIVILIAN_AIRCRAFT":
        pos_z, v_mag = np.random.uniform(7000, 11000), np.random.uniform(220, 260)
    elif active_class == "FIGHTER_AIRCRAFT":
        pos_z, v_mag = np.random.uniform(500, 15000), np.random.uniform(250, 750)
    elif active_class == "HELICOPTER":
        pos_z, v_mag = np.random.uniform(20, 1500), np.random.uniform(0, 85)
    elif active_class == "DRONE":
        pos_z, v_mag = np.random.uniform(20, 3000), np.random.uniform(5, 60)
    elif active_class == "SURFACE_MISSILE":
        pos_z, v_mag = np.random.uniform(15, 100), np.random.uniform(300, 800)
    elif active_class == "AERIAL_MISSILE":
        pos_z, v_mag = np.random.uniform(2000, 12000), np.random.uniform(600, 1200)
    elif active_class == "SKIFF":
        pos_z, v_mag = 0.0, np.random.uniform(10, 28)
    elif active_class == "BOAT":
        pos_z, v_mag = 0.0, np.random.uniform(4, 18)
    elif active_class == "SHIP":
        pos_z, v_mag = 0.0, np.random.uniform(2, 10)

    angle = random.uniform(0, 2 * np.pi)
    vel = np.array([v_mag * np.cos(angle), v_mag * np.sin(angle), 0.0])

    start_time = 3991182426
    actual_elapsed_time = 0.0

    maneuver_timer = 0
    maneuver_type = "NONE"
    target_turn_rate = 0.0
    in_burst_loss = False

    for t in range(duration):
        # ---------------------------------------------------
        # 3. DISGUISE BREAKOUT LOGIC
        # ---------------------------------------------------
        if is_disguised and t >= disguise_breakout_step:
            if track_class == "FIGHTER_AIRCRAFT":
                vel *= 1.8  # Supersonic burst
                vel[2] -= 150 # Drop altitude rapidly
            elif track_class == "DRONE":
                angle += np.radians(90) # Physically impossible instant 90-deg pivot
                vel[0], vel[1] = v_mag * np.cos(angle), v_mag * np.sin(angle)
            elif track_class == "SURFACE_MISSILE":
                vel[2] += 250 # Terminal pop-up climb

            is_disguised = False
            active_class = track_class # Revert state machine to true capabilities
            maneuver_timer = 0 # Cancel current disguised maneuver

        # ---------------------------------------------------
        # 4. SUSTAINED MANEUVER LOGIC (By Active Class)
        # ---------------------------------------------------
        if maneuver_timer <= 0:
            if active_class == "FIGHTER_AIRCRAFT" and random.random() < 0.15:
                maneuver_timer = random.randint(4, 10)
                maneuver_type = random.choice(["EVASIVE_TURN", "SPLIT_S_DIVE"])
                target_turn_rate = np.radians(random.uniform(-15, 15))
            elif active_class == "CIVILIAN_AIRCRAFT" and random.random() < 0.10:
                maneuver_timer = random.randint(15, 30)
                maneuver_type = "HOLDING_PATTERN"
                target_turn_rate = np.radians(random.choice([-3, 3]))
            elif active_class in ["HELICOPTER", "DRONE"] and random.random() < 0.20:
                maneuver_timer = random.randint(3, 8)
                maneuver_type = random.choice(["POP_UP", "SUDDEN_HOVER"])
            elif active_class in ["SKIFF", "BOAT"] and random.random() < 0.25:
                maneuver_timer = random.randint(2, 6)
                maneuver_type = "ZIG_ZAG"
                target_turn_rate = np.radians(random.uniform(-25, 25))
            elif active_class in ["SURFACE_MISSILE", "AERIAL_MISSILE"] and random.random() < 0.10:
                maneuver_timer = random.randint(3, 5)
                maneuver_type = "TERMINAL_POP_UP"
            else:
                maneuver_type = "CRUISE"

        if maneuver_timer > 0:
            maneuver_timer -= 1
            if maneuver_type in ["EVASIVE_TURN", "HOLDING_PATTERN", "ZIG_ZAG"]:
                c, s = np.cos(target_turn_rate * dt), np.sin(target_turn_rate * dt)
                vel[0], vel[1] = vel[0]*c - vel[1]*s, vel[0]*s + vel[1]*c
            elif maneuver_type == "SPLIT_S_DIVE":
                vel[2] -= 40 * dt
                vel[0] *= 1.05; vel[1] *= 1.05
            elif maneuver_type == "POP_UP":
                vel[2] = 25
                vel[0] *= 0.5; vel[1] *= 0.5
            elif maneuver_type == "SUDDEN_HOVER":
                vel *= 0.5
            elif maneuver_type == "TERMINAL_POP_UP":
                vel[2] = 50
        else:
            if active_class not in ["CIVILIAN_AIRCRAFT", "SHIP"]:
                vel += np.random.normal(0, 1.5, 3)
            else:
                vel += np.random.normal(0, 0.2, 3)
                vel[2] *= 0.95

        # ---------------------------------------------------
        # 5. PHYSICS & BOUNDARY UPDATE
        # ---------------------------------------------------
        if active_class in ["SKIFF", "BOAT", "SHIP"]: vel[2] = 0.0

        pos_x += vel[0] * dt
        pos_y += vel[1] * dt
        pos_z += vel[2] * dt

        if pos_z < 0 and active_class not in ["SKIFF", "BOAT", "SHIP"]:
            pos_z, vel[2] = 0.0, 0.0
        if active_class in ["SKIFF", "BOAT", "SHIP"]: pos_z = 0.0

        actual_elapsed_time += dt

        # ---------------------------------------------------
        # 6. BURST PACKET LOSS SIMULATION
        # ---------------------------------------------------
        if in_burst_loss:
            if random.random() < 0.60:
                in_burst_loss = False
            else:
                continue
        else:
            if random.random() < 0.08:
                in_burst_loss = True
                continue

        # ---------------------------------------------------
        # 7. RANGE-DEPENDENT SENSOR NOISE
        # ---------------------------------------------------
        r_mag = np.sqrt(pos_x**2 + pos_y**2 + pos_z**2)
        noise_multiplier = max(1.0, (r_mag / 25000) ** 1.5)

        pos_x_noisy = pos_x + np.random.normal(0, 15 * noise_multiplier)
        pos_y_noisy = pos_y + np.random.normal(0, 15 * noise_multiplier)
        pos_z_noisy = abs(np.random.normal(0, 8 * noise_multiplier)) if active_class in ["SKIFF", "BOAT", "SHIP"] else pos_z + np.random.normal(0, 20 * noise_multiplier)

        vel_x_noisy = vel[0] + np.random.normal(0, 1.5 * noise_multiplier)
        vel_y_noisy = vel[1] + np.random.normal(0, 1.5 * noise_multiplier)
        vel_z_noisy = vel[2] + np.random.normal(0, 1.5 * noise_multiplier)

        timestamp = start_time + int(actual_elapsed_time)

        pos_x_list.append(pos_x_noisy)
        pos_y_list.append(pos_y_noisy)
        pos_z_list.append(pos_z_noisy)
        vel_x_list.append(vel_x_noisy)
        vel_y_list.append(vel_y_noisy)
        vel_z_list.append(vel_z_noisy)
        track_no_list.append(track_id)
        track_class_list.append(track_class) # Record the TRUE class for evaluation
        ntp_sec_list.append(timestamp)

    return pd.DataFrame({
        'pos_x': pos_x_list, 'pos_y': pos_y_list, 'pos_z': pos_z_list,
        'vel_x': vel_x_list, 'vel_y': vel_y_list, 'vel_z': vel_z_list,
        'track_no': track_no_list, 'track_class': track_class_list, 'ntp_sec': ntp_sec_list
    })

def generate_stress_test_dataset(num_samples=270, duration=80):
    dfs = []
    samples_per_class = num_samples // len(CLASSES)

    for class_choice in CLASSES:
        for _ in range(samples_per_class):
            df = generate_stress_trajectory(class_choice, duration=duration)
            if df is not None and len(df) > 5:
                dfs.append(df)

    return pd.concat(dfs, ignore_index=True)

if __name__ == "__main__":
    print("Initializing 9-Class Real-World Radar Stress Test Generator...")
    # Generating tracks, tracked over 80 potential scans
    test_df = generate_stress_test_dataset(num_samples=270, duration=80)
    test_df.to_csv("test_radar_data_9class.csv", index=False)

    print(f"Successfully generated advanced stress-test dataset with {len(test_df)} packets.")
    print("Features injected:")
    print(" - Burst packet loss (Time jumps > 10s)")

    test_df['dt'] = test_df.groupby('track_no')['ntp_sec'].diff()
    max_jump = test_df['dt'].max()

    print(f" - Maximum sensor blackout period detected: {max_jump} seconds")
    print(" - Range-dependent exponentially scaling noise")
    print(" - Sustained Holding Patterns, Evasive Dives, and Hovering Maneuvers")
    print(" - Disguise scenarios (Tactical targets masking as civilian/commercial craft before breakout)")
```

```python
import pandas as pd
import numpy as np
import time
from tensorflow.keras.models import load_model
import joblib
import warnings
warnings.filterwarnings('ignore')

# --- SMART PREPROCESSING BRIDGE FOR 9-CLASS MODEL ---
def robust_feature_extractor(df):
    """Calculates all advanced physics features dynamically from raw telemetry packets."""
    df = df.copy()

    if 'track_no' not in df.columns: df['track_no'] = 1
    if 'ntp_sec' not in df.columns: df['ntp_sec'] = np.arange(len(df)) * 2.0

    df = df.sort_values(['track_no', 'ntp_sec']).reset_index(drop=True)
    df['dt'] = df.groupby('track_no')['ntp_sec'].diff().fillna(2.0).replace(0, 2.0)

    # 1. Base Kinematics
    for axis in ['x', 'y', 'z']:
        if f'vel_{axis}' not in df.columns:
            df[f'vel_{axis}'] = df.groupby('track_no')[f'pos_{axis}'].diff() / df['dt']
            df[f'vel_{axis}'] = df[f'vel_{axis}'].fillna(0)

    for axis in ['x', 'y', 'z']:
        if f'accel_{axis}' not in df.columns:
            df[f'accel_{axis}'] = df.groupby('track_no')[f'vel_{axis}'].diff() / df['dt']
            df[f'accel_{axis}'] = df[f'accel_{axis}'].fillna(0)

    df['accel_mag'] = np.sqrt(df['accel_x']**2 + df['accel_y']**2 + df['accel_z']**2)

    df['jerk_x'] = df.groupby('track_no')['accel_x'].diff() / df['dt']
    df['jerk_y'] = df.groupby('track_no')['accel_y'].diff() / df['dt']
    df['jerk_z'] = df.groupby('track_no')['accel_z'].diff() / df['dt']
    df[['jerk_x', 'jerk_y', 'jerk_z']] = df[['jerk_x', 'jerk_y', 'jerk_z']].fillna(0)
    df['jerk_mag'] = np.sqrt(df['jerk_x']**2 + df['jerk_y']**2 + df['jerk_z']**2)

    if 'vel_mag' not in df.columns: df['vel_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2 + df['vel_z']**2)
    df['vel_xy_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2)
    if 'range_mag' not in df.columns: df['range_mag'] = np.sqrt(df['pos_x']**2 + df['pos_y']**2 + df['pos_z']**2)

    df['jerk_to_accel_ratio'] = df['jerk_mag'] / (df['accel_mag'] + 1e-5)

    # 2. 3D Differential Geometry (Curvature & Torsion)
    v = df[['vel_x', 'vel_y', 'vel_z']].values
    a = df[['accel_x', 'accel_y', 'accel_z']].values
    j = df[['jerk_x', 'jerk_y', 'jerk_z']].values

    v_cross_a = np.cross(v, a)
    v_cross_a_mag = np.linalg.norm(v_cross_a, axis=1)
    v_mag = df['vel_mag'].values

    curvature = v_cross_a_mag / (v_mag**3 + 1e-6)
    df['curvature_3d'] = np.nan_to_num(curvature)

    dot_v_cross_a_j = np.einsum('ij,ij->i', v_cross_a, j)
    torsion = dot_v_cross_a_j / (v_cross_a_mag**2 + 1e-6)
    df['torsion_3d'] = np.nan_to_num(torsion)

    # 3. Energy Dynamics
    g = 9.81
    df['specific_energy'] = g * df['pos_z'] + 0.5 * (df['vel_mag']**2)
    df['energy_rate'] = g * df['vel_z'] + (df['vel_x']*df['accel_x'] + df['vel_y']*df['accel_y'] + df['vel_z']*df['accel_z'])

    df['z_ratio'] = df['vel_z'] / (df['vel_mag'] + 1e-8)
    df['vertical_accel_ratio'] = df['accel_z'] / (df['accel_mag'] + 1e-8)

    # 4. Live Rolling Stats (Captures transient breakouts & noise)
    for col in ['accel_mag', 'jerk_mag', 'energy_rate']:
        df[f'{col}_roll_mean'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).mean())
        df[f'{col}_roll_std'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).std()).fillna(0.0)

    # 5. Live Cumulative Tracking Extremes
    df['cum_max_accel'] = df.groupby('track_no')['accel_mag'].cummax()
    df['cum_max_jerk'] = df.groupby('track_no')['jerk_mag'].cummax()
    df['cum_max_energy_rate'] = df.groupby('track_no')['energy_rate'].cummax()

    df = df.fillna(0.0)

    # Strict adherence to the 22 features required by the Transformer model
    required_features = [
        'pos_z', 'vel_mag', 'vel_xy_mag', 'range_mag', 'accel_mag', 'jerk_mag',
        'jerk_to_accel_ratio', 'curvature_3d', 'torsion_3d', 'specific_energy',
        'energy_rate', 'z_ratio', 'vertical_accel_ratio',
        'accel_mag_roll_mean', 'accel_mag_roll_std',
        'jerk_mag_roll_mean', 'jerk_mag_roll_std',
        'energy_rate_roll_mean', 'energy_rate_roll_std',
        'cum_max_accel', 'cum_max_jerk', 'cum_max_energy_rate'
    ]

    for col in required_features:
        if col not in df.columns:
            df[col] = 0.0

    return df

# --- STATEFUL TRACK MANAGER ---
class LiveTrackManager:
    def __init__(self, model_path="trained_model/radar_target_classifier.keras"):
        print("Loading 9-Class Transformer-Hybrid Model...")
        self.model = load_model(model_path, compile=False)
        self.le = joblib.load("trained_model/label_encoder.pkl")
        self.scaler = joblib.load("trained_model/feature_scaler.pkl")

        # Memory buffer for live tracks
        self.active_tracks = {}

        self.feature_cols = [
            'pos_z', 'vel_mag', 'vel_xy_mag', 'range_mag', 'accel_mag', 'jerk_mag',
            'jerk_to_accel_ratio', 'curvature_3d', 'torsion_3d', 'specific_energy',
            'energy_rate', 'z_ratio', 'vertical_accel_ratio',
            'accel_mag_roll_mean', 'accel_mag_roll_std',
            'jerk_mag_roll_mean', 'jerk_mag_roll_std',
            'energy_rate_roll_mean', 'energy_rate_roll_std',
            'cum_max_accel', 'cum_max_jerk', 'cum_max_energy_rate'
        ]

        self.seq_length = 40  # Updated to match the new model architecture

    def process_packet(self, raw_packet):
        """
        Receives a single dictionary (representing a UDP radar packet).
        Returns the updated prediction for that track.
        """
        tid = raw_packet['track_no']

        # 1. Update Track Memory
        if tid not in self.active_tracks:
            self.active_tracks[tid] = []

        self.active_tracks[tid].append(raw_packet)

        # Buffer limits (Keeps RAM usage low while maintaining enough history for cumulative features)
        if len(self.active_tracks[tid]) > 200:
            self.active_tracks[tid] = self.active_tracks[tid][-200:]

        # 2. Extract Features dynamically
        track_df = pd.DataFrame(self.active_tracks[tid])
        features_df = robust_feature_extractor(track_df)

        # 3. Predict on the newly generated sequence
        try:
            features = features_df[self.feature_cols].values
            features_scaled = self.scaler.transform(features)

            # Pad or truncate to exact sequence length
            if len(features_scaled) >= self.seq_length:
                input_data = features_scaled[-self.seq_length:].reshape(1, self.seq_length, -1)
            else:
                padded = np.pad(features_scaled, ((self.seq_length - len(features_scaled), 0), (0, 0)), 'constant')
                input_data = padded.reshape(1, self.seq_length, -1)

            pred_probs = self.model.predict(input_data, verbose=0)[0]
            predicted_class_idx = np.argmax(pred_probs)
            confidence = np.max(pred_probs)
            predicted_label = self.le.inverse_transform([predicted_class_idx])[0]

            packet_count = len(self.active_tracks[tid])
            return {'track_id': tid, 'prediction': predicted_label, 'confidence': float(confidence), 'packets_received': packet_count}

        except Exception as e:
            return {'track_id': tid, 'prediction': f'ERROR: {str(e)}', 'confidence': 0.0, 'packets_received': 0}

if __name__ == "__main__":
    # Point to the newly generated 9-class stress test data
    TEST_FILE_PATH = "/content/radar_data.csv"

    manager = LiveTrackManager()

    print(f"\nSimulating live packet stream from {TEST_FILE_PATH}...")
    raw_stream_df = pd.read_csv(TEST_FILE_PATH).dropna(subset=['track_no'])

    # Sort purely by time to completely scramble the track numbers, mimicking a multiplexed feed
    raw_stream_df = raw_stream_df.sort_values('ntp_sec')
    live_queue = raw_stream_df.to_dict('records')

    print("\n" + "="*75)
    print(" LIVE MULTIPLEXED MULTI-CLASS DATA FEED SIMULATION")
    print("="*75)

    # Let's watch targets from 3 different classes to observe the behavior
    sampled_tracks = raw_stream_df.drop_duplicates(subset=['track_no'])
    target_ids_to_watch = sampled_tracks['track_no'].head(4).tolist()

    for packet in live_queue:
        result = manager.process_packet(packet)

        if result['track_id'] in target_ids_to_watch:
            actual_class = packet.get('track_class', 'UNKNOWN')

            # Formatting logic for terminal readability
            if result['prediction'] == actual_class: status = "✅"
            elif actual_class == "UNKNOWN": status = "🔍"
            else: status = "⚠️ (Disguised/Pending)"

            print(f"📡 [Time: {packet['ntp_sec']}] Track {result['track_id']} (Actual True Target: {actual_class})")
            print(f"   -> Packets buffered: {result['packets_received']:02d} | Output: {result['prediction']:<18s} | Confidence: {result['confidence']:.2%} {status}\n")
```

```python
import pandas as pd
import numpy as np
import joblib
from tensorflow.keras.models import load_model
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings
warnings.filterwarnings('ignore')

# 1. Reuse your exact feature extractor (Keep it identical to your live script)
def robust_feature_extractor(df):
    df = df.copy()
    if 'track_no' not in df.columns: df['track_no'] = 1
    if 'ntp_sec' not in df.columns: df['ntp_sec'] = np.arange(len(df)) * 2.0
    df = df.sort_values(['track_no', 'ntp_sec']).reset_index(drop=True)
    df['dt'] = df.groupby('track_no')['ntp_sec'].diff().fillna(2.0).replace(0, 2.0)

    for axis in ['x', 'y', 'z']:
        if f'vel_{axis}' not in df.columns:
            df[f'vel_{axis}'] = df.groupby('track_no')[f'pos_{axis}'].diff() / df['dt']
            df[f'vel_{axis}'] = df[f'vel_{axis}'].fillna(0)
        if f'accel_{axis}' not in df.columns:
            df[f'accel_{axis}'] = df.groupby('track_no')[f'vel_{axis}'].diff() / df['dt']
            df[f'accel_{axis}'] = df[f'accel_{axis}'].fillna(0)

    df['accel_mag'] = np.sqrt(df['accel_x']**2 + df['accel_y']**2 + df['accel_z']**2)
    df['jerk_x'] = df.groupby('track_no')['accel_x'].diff() / df['dt']
    df['jerk_y'] = df.groupby('track_no')['accel_y'].diff() / df['dt']
    df['jerk_z'] = df.groupby('track_no')['accel_z'].diff() / df['dt']
    df[['jerk_x', 'jerk_y', 'jerk_z']] = df[['jerk_x', 'jerk_y', 'jerk_z']].fillna(0)
    df['jerk_mag'] = np.sqrt(df['jerk_x']**2 + df['jerk_y']**2 + df['jerk_z']**2)

    if 'vel_mag' not in df.columns: df['vel_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2 + df['vel_z']**2)
    df['vel_xy_mag'] = np.sqrt(df['vel_x']**2 + df['vel_y']**2)
    if 'range_mag' not in df.columns: df['range_mag'] = np.sqrt(df['pos_x']**2 + df['pos_y']**2 + df['pos_z']**2)

    df['jerk_to_accel_ratio'] = df['jerk_mag'] / (df['accel_mag'] + 1e-5)

    v = df[['vel_x', 'vel_y', 'vel_z']].values
    a = df[['accel_x', 'accel_y', 'accel_z']].values
    j = df[['jerk_x', 'jerk_y', 'jerk_z']].values

    v_cross_a = np.cross(v, a)
    v_cross_a_mag = np.linalg.norm(v_cross_a, axis=1)
    df['curvature_3d'] = np.nan_to_num(v_cross_a_mag / (df['vel_mag'].values**3 + 1e-6))
    df['torsion_3d'] = np.nan_to_num(np.einsum('ij,ij->i', v_cross_a, j) / (v_cross_a_mag**2 + 1e-6))

    g = 9.81
    df['specific_energy'] = g * df['pos_z'] + 0.5 * (df['vel_mag']**2)
    df['energy_rate'] = g * df['vel_z'] + (df['vel_x']*df['accel_x'] + df['vel_y']*df['accel_y'] + df['vel_z']*df['accel_z'])
    df['z_ratio'] = df['vel_z'] / (df['vel_mag'] + 1e-8)
    df['vertical_accel_ratio'] = df['accel_z'] / (df['accel_mag'] + 1e-8)

    for col in ['accel_mag', 'jerk_mag', 'energy_rate']:
        df[f'{col}_roll_mean'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).mean())
        df[f'{col}_roll_std'] = df.groupby('track_no')[col].transform(lambda x: x.rolling(5, min_periods=1).std()).fillna(0.0)

    df['cum_max_accel'] = df.groupby('track_no')['accel_mag'].cummax()
    df['cum_max_jerk'] = df.groupby('track_no')['jerk_mag'].cummax()
    df['cum_max_energy_rate'] = df.groupby('track_no')['energy_rate'].cummax()

    required_features = [
        'pos_z', 'vel_mag', 'vel_xy_mag', 'range_mag', 'accel_mag', 'jerk_mag',
        'jerk_to_accel_ratio', 'curvature_3d', 'torsion_3d', 'specific_energy',
        'energy_rate', 'z_ratio', 'vertical_accel_ratio',
        'accel_mag_roll_mean', 'accel_mag_roll_std',
        'jerk_mag_roll_mean', 'jerk_mag_roll_std',
        'energy_rate_roll_mean', 'energy_rate_roll_std',
        'cum_max_accel', 'cum_max_jerk', 'cum_max_energy_rate'
    ]
    for col in required_features:
        if col not in df.columns: df[col] = 0.0
    return df.fillna(0.0), required_features


def generate_executive_dashboard(csv_path, track_id_to_plot=None):
    print("Loading Models and Data for Analysis...")
    model = load_model("trained_model/radar_target_classifier.keras", compile=False)
    le = joblib.load("trained_model/label_encoder.pkl")
    scaler = joblib.load("trained_model/feature_scaler.pkl")
    seq_length = 40

    df_full = pd.read_csv(csv_path).dropna(subset=['track_no'])

    # Auto-select a target if none provided (Preferably a Fighter or Missile for the disguise effect)
    if track_id_to_plot is None:
        interesting_classes = ["FIGHTER_AIRCRAFT", "SURFACE_MISSILE", "DRONE"]
        sample = df_full[df_full['track_class'].isin(interesting_classes)]
        if not sample.empty:
            track_id_to_plot = sample['track_no'].iloc[0]
        else:
            track_id_to_plot = df_full['track_no'].iloc[0]

    target_df = df_full[df_full['track_no'] == track_id_to_plot].sort_values('ntp_sec')
    actual_class = target_df['track_class'].iloc[0]

    print(f"Simulating history for Track {track_id_to_plot} (Actual: {actual_class})...")

    history_times, history_probs, history_preds = [], [], []
    active_buffer = []

    # Simulate streaming packet-by-packet to get real-time model confidence changes
    for _, row in target_df.iterrows():
        active_buffer.append(row.to_dict())
        current_df = pd.DataFrame(active_buffer)

        features_df, feature_cols = robust_feature_extractor(current_df)
        features_scaled = scaler.transform(features_df[feature_cols].values)

        if len(features_scaled) >= seq_length:
            input_data = features_scaled[-seq_length:].reshape(1, seq_length, -1)
        else:
            padded = np.pad(features_scaled, ((seq_length - len(features_scaled), 0), (0, 0)), 'constant')
            input_data = padded.reshape(1, seq_length, -1)

        probs = model.predict(input_data, verbose=0)[0]
        history_times.append(row['ntp_sec'])
        history_probs.append(probs)
        history_preds.append(le.inverse_transform([np.argmax(probs)])[0])

    probs_df = pd.DataFrame(history_probs, columns=le.classes_)

    # ==========================================
    # BUILD INTERACTIVE PLOTLY DASHBOARD
    # ==========================================
    print("Generating Interactive HTML Dashboard...")
    fig = make_subplots(
        rows=2, cols=2,
        column_widths=[0.6, 0.4],
        row_heights=[0.5, 0.5],
        specs=[[{"type": "scatter3d", "rowspan": 2}, {"type": "scatter"}],
               [None, {"type": "scatter"}]],
        subplot_titles=("3D Radar Trajectory (Hover for metrics)",
                        "AI Confidence Over Time (Disguise Breakout)",
                        "Kinematic Energy Signature (The 'Tell')")
    )

    # 1. 3D Trajectory Plot
    fig.add_trace(go.Scatter3d(
        x=target_df['pos_x'], y=target_df['pos_y'], z=target_df['pos_z'],
        mode='lines+markers',
        marker=dict(
            size=4,
            color=target_df['vel_mag'], # Color by speed
            colorscale='Turbo',
            colorbar=dict(title="Speed (m/s)", x=-0.1),
            showscale=True
        ),
        line=dict(width=2),
        text=[f"Time: {t}<br>Speed: {v:.1f}m/s<br>Pred: {p}"
              for t, v, p in zip(history_times, target_df['vel_mag'], history_preds)],
        hoverinfo="text",
        name="Flight Path"
    ), row=1, col=1)

    # 2. Probability / Confidence Timeline
    for class_name in le.classes_:
        # Only plot classes that the model actually considered (prob > 5%) to keep it clean
        if probs_df[class_name].max() > 0.05:
            fig.add_trace(go.Scatter(
                x=history_times, y=probs_df[class_name],
                mode='lines',
                name=f"Prob: {class_name}",
                line_shape='spline',
                fill='tozeroy' if class_name == actual_class else 'none'
            ), row=1, col=2)

    # 3. Kinematic Energy Signature (Specific Energy Rate)
    # This proves WHY the model changed its mind
    full_features, _ = robust_feature_extractor(target_df)
    fig.add_trace(go.Scatter(
        x=history_times, y=full_features['energy_rate'],
        mode='lines',
        name='Specific Energy Rate (W/kg)',
        line=dict(color='red', width=2)
    ), row=2, col=2)

    fig.add_trace(go.Scatter(
        x=history_times, y=full_features['accel_mag'],
        mode='lines',
        name='Acceleration (m/s²)',
        line=dict(color='orange', width=2, dash='dash')
    ), row=2, col=2)

    # Styling and Layout
    fig.update_layout(
        title=f"Target Intelligence Report: Track {track_id_to_plot} | True Class: {actual_class}",
        height=900,
        width=1600,
        template="plotly_dark",
        showlegend=True,
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=1.05)
    )

    # Auto-open in the default web browser
    output_file = f"target_analysis_{track_id_to_plot}.html"
    fig.write_html(output_file, auto_open=True)
    print(f"Dashboard saved and opened: {output_file}")

if __name__ == "__main__":
    # Point this to your 9-class stress test data
    TEST_DATA_PATH = "/content/radar_data.csv"

    # You can pass a specific track_no here, or leave as None to auto-pick an interesting one
    generate_executive_dashboard(TEST_DATA_PATH, track_id_to_plot=None)
```
